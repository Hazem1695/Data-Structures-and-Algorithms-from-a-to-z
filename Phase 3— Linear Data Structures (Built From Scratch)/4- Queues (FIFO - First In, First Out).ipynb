# **4: Queues (FIFO - First In, First Out)**

## 1: Theory

**What problem does this structure solve?**
While a stack manages operations where the newest data takes priority, many real-world systems require treating data with absolute fairness: **first come, first served**. For example:

* **Print Jobs:** Multiple computers send documents to a single printer; the first document sent must be printed first.
* **Web Server Requests:** Incoming traffic spikes require a buffer to hold requests so they can be processed in the exact chronological order they arrived.

A **Queue** is a linear data structure governed by the **First In, First Out (FIFO)** principle. It permits insertions exclusively at one end (the **Rear/Tail**) and removals exclusively from the opposite end (the **Front/Head**).

#### **The Memory Bottleneck Problem**

If you implement a queue naively using a standard dynamic array, an architectural bottleneck arises:

* **Enqueue (Insertion at Rear):** Appending to the end of an array is highly efficient, taking an amortized $O(1)$ time.
* **Dequeue (Removal from Front):** Removing the element at index 0 forces every single remaining element in the array to shift left by one slot. This turns your dequeue operation into an expensive $O(N)$ time penalty. For a queue handling millions of network packets, this causes massive execution delays.

To eliminate this memory bottleneck, we use two pointer-based architectures that guarantee true **$O(1)$ time for both operations**:

1. **Linked List Implementation:** We maintain both a `head` pointer and a `tail` pointer. We enqueue at the tail and dequeue at the head, requiring no data shifting whatsoever.
2. **Circular Array (Ring Buffer) Implementation:** We use a fixed-size array, but instead of shifting elements left, we allow the `head` and `tail` tracking variables to "wrap around" to the beginning of the array using modulo arithmetic.

````python
class QueueArray:
    # Define a queue using a Python list

    def __init__(self):
        # Initialize an empty list to store queue elements
        self.items = []
        # Keep track of the index of the front element
        self.front = 0

    def enqueue(self, value):
        # Add a new element to the end (rear) of the queue
        self.items.append(value)

    def dequeue(self):
        # Remove and return the front element of the queue
        if self.is_empty():
            # If queue is empty, raise an exception
            raise Exception("Queue Empty")

        # Get the current front element
        value = self.items[self.front]
        # Move the front pointer one step forward
        self.front += 1

        # Return the removed element
        return value

    def peek(self):
        # Return the front element without removing it
        if self.is_empty():
            # If queue is empty, raise an exception
            raise Exception("Queue Empty")

        # Return the element at the front index
        return self.items[self.front]

    def is_empty(self):
        # Check if the queue is empty
        # Queue is empty when front index passes the last element
        return self.front >= len(self.items)

    def size(self):
        # Return the number of elements currently in the queue
        # Subtract front index from total length to get active size
        return len(self.items) - self.front
````

````python
# Test code for QueueArray

# Create a new queue
queue = QueueArray()

# Enqueue some elements
print("Enqueue 10")
queue.enqueue(10)
print("Enqueue 20")
queue.enqueue(20)
print("Enqueue 30")
queue.enqueue(30)

# Check size
print("Current size:", queue.size())  # Expected: 3

# Peek at the front element
print("Peek front:", queue.peek())    # Expected: 10

# Dequeue elements one by one
print("Dequeue:", queue.dequeue())    # Expected: 10
print("Dequeue:", queue.dequeue())    # Expected: 20

# Check size again
print("Current size:", queue.size())  # Expected: 1

# Peek at the new front
print("Peek front:", queue.peek())    # Expected: 30

# Dequeue last element
print("Dequeue:", queue.dequeue())    # Expected: 30

# Check if queue is empty
print("Is empty?", queue.is_empty())  # Expected: True

# Try to peek or dequeue from empty queue (will raise Exception)
try:
    print("Peek front:", queue.peek())
except Exception as e:
    print("Error:", e)

try:
    print("Dequeue:", queue.dequeue())
except Exception as e:
    print("Error:", e)
````

### **Queue Using Linked List**
```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class QueueLinkedList:

    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0

    def enqueue(self, value):
        new_node = Node(value)
        if self.rear is None:
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node
        self.count += 1

    def dequeue(self):
        if self.front is None:
            raise Exception("Queue Empty")
        value = self.front.data
        self.front = self.front.next
        if self.front is None:
            self.rear = None
        self.count -= 1
        return value

    def peek(self):
        if self.front is None:
            raise Exception("Queue Empty")
        return self.front.data
    def is_empty(self):
        return self.front is None

    def size(self):
        return self.count
```

````python
# Node class for linked list
class Node:
    def __init__(self, data):
        self.data = data      # Store the value
        self.next = None      # Pointer to the next node

# Queue implemented using a linked list
class QueueLinkedList:
    def __init__(self):
        self.front = None     # Points to the first element (front of queue)
        self.rear = None      # Points to the last element (rear of queue)
        self.count = 0        # Keep track of number of elements

    def enqueue(self, value):
        # Add a new element at the rear of the queue
        new_node = Node(value)
        if self.rear is None:         # If queue is empty
            self.front = new_node     # Front and rear both point to new node
            self.rear = new_node
        else:
            self.rear.next = new_node # Link current rear to new node
            self.rear = new_node      # Update rear pointer
        self.count += 1               # Increase size counter

    def dequeue(self):
        # Remove and return the front element
        if self.front is None:
            raise Exception("Queue Empty")
        value = self.front.data       # Get value at front
        self.front = self.front.next  # Move front pointer forward
        if self.front is None:        # If queue becomes empty
            self.rear = None          # Reset rear pointer too
        self.count -= 1               # Decrease size counter
        return value

    def peek(self):
        # Return the front element without removing it
        if self.front is None:
            raise Exception("Queue Empty")
        return self.front.data

    def is_empty(self):
        # Check if queue has no elements
        return self.front is None

    def size(self):
        # Return number of elements in queue
        return self.count
````

````python
# Test code for QueueLinkedList

queue = QueueLinkedList()

# Enqueue elements
print("Enqueue 10")
queue.enqueue(10)
print("Enqueue 20")
queue.enqueue(20)
print("Enqueue 30")
queue.enqueue(30)

# Check size
print("Current size:", queue.size())  # Expected: 3

# Peek at front
print("Peek front:", queue.peek())    # Expected: 10

# Dequeue elements
print("Dequeue:", queue.dequeue())    # Expected: 10
print("Dequeue:", queue.dequeue())    # Expected: 20

# Check size again
print("Current size:", queue.size())  # Expected: 1

# Peek at new front
print("Peek front:", queue.peek())    # Expected: 30

# Dequeue last element
print("Dequeue:", queue.dequeue())    # Expected: 30

# Check if empty
print("Is empty?", queue.is_empty())  # Expected: True

# Try to peek/dequeue from empty queue
try:
    print("Peek front:", queue.peek())
except Exception as e:
    print("Error:", e)

try:
    print("Dequeue:", queue.dequeue())
except Exception as e:
    print("Error:", e)
````

Problem 4

Implement a queue using a singly linked list.

````python
# Node class for singly linked list
class Node:
    def __init__(self, data):
        self.data = data      # Store the value
        self.next = None      # Pointer to the next node


# Queue implemented using singly linked list
class QueueLinkedList:
    def __init__(self):
        self.front = None     # Points to the first element (front of queue)
        self.rear = None      # Points to the last element (rear of queue)
        self.count = 0        # Number of elements in the queue

    def enqueue(self, value):
        # Add a new element at the rear
        new_node = Node(value)

        if self.rear is None:         # If queue is empty
            self.front = new_node     # Both front and rear point to new node
            self.rear = new_node
        else:
            self.rear.next = new_node # Link current rear to new node
            self.rear = new_node      # Update rear pointer

        self.count += 1               # Increase size counter

    def dequeue(self):
        # Remove and return the front element
        if self.front is None:
            raise Exception("Queue Empty")

        value = self.front.data       # Get value at front
        self.front = self.front.next  # Move front pointer forward

        if self.front is None:        # If queue becomes empty
            self.rear = None          # Reset rear pointer

        self.count -= 1               # Decrease size counter
        return value

    def peek(self):
        # Return the front element without removing it
        if self.front is None:
            raise Exception("Queue Empty")
        return self.front.data

    def is_empty(self):
        # Check if queue has no elements
        return self.front is None

    def size(self):
        # Return number of elements in queue
        return self.count

    def show(self):
        # Display all elements from front → rear
        current = self.front
        print("Queue contents:", end=" ")
        while current:
            print(current.data, end=" ")
            current = current.next
        print()
````

````python
# Create a queue
queue = QueueLinkedList()

# Enqueue elements
queue.enqueue(10)
queue.enqueue(20)
queue.enqueue(30)

# Show queue
queue.show()   # Expected: Queue contents: 10 20 30

# Peek at front
print("Peek front:", queue.peek())    # Expected: 10

# Dequeue elements
print("Dequeue:", queue.dequeue())    # Expected: 10
print("Dequeue:", queue.dequeue())    # Expected: 20

# Show queue again
queue.show()   # Expected: Queue contents: 30

# Check size
print("Current size:", queue.size())  # Expected: 1

# Dequeue last element
print("Dequeue:", queue.dequeue())    # Expected: 30

# Check if empty
print("Is empty?", queue.is_empty())  # Expected: True
````

Problem 5

Implement a queue using two stacks.

Hint:
```
Stack A
Stack B
```
Classic interview problem.


````python
class QueueTwoStacks:
    def __init__(self):
        self.stackA = []   # First stack (for enqueue)
        self.stackB = []   # Second stack (for dequeue)

    def enqueue(self, value):
        # Always push new elements onto stackA
        self.stackA.append(value)

    def dequeue(self):
        # If stackB is empty, transfer all elements from stackA to stackB
        if not self.stackB:
            while self.stackA:
                self.stackB.append(self.stackA.pop())

        # If stackB is still empty, queue is empty
        if not self.stackB:
            raise Exception("Queue Empty")

        # Pop from stackB (front of queue)
        return self.stackB.pop()

    def peek(self):
        # Ensure stackB has the front element
        if not self.stackB:
            while self.stackA:
                self.stackB.append(self.stackA.pop())

        if not self.stackB:
            raise Exception("Queue Empty")

        return self.stackB[-1]

    def is_empty(self):
        # Queue is empty if both stacks are empty
        return not self.stackA and not self.stackB

    def size(self):
        # Total size is sum of both stacks
        return len(self.stackA) + len(self.stackB)

    def show(self):
        # Display queue contents in correct order (front → rear)
        temp = self.stackB[::-1] + self.stackA
        print("Queue contents:", temp)
````

````python
queue = QueueTwoStacks()

# Enqueue elements
queue.enqueue(10)
queue.enqueue(20)
queue.enqueue(30)

# Show queue
queue.show()   # Expected: Queue contents: [10, 20, 30]

# Peek front
print("Peek front:", queue.peek())    # Expected: 10

# Dequeue elements
print("Dequeue:", queue.dequeue())    # Expected: 10
print("Dequeue:", queue.dequeue())    # Expected: 20

# Show queue again
queue.show()   # Expected: Queue contents: [30]

# Check size
print("Current size:", queue.size())  # Expected: 1

# Dequeue last element
print("Dequeue:", queue.dequeue())    # Expected: 30

# Check if empty
print("Is empty?", queue.is_empty())  # Expected: True
````

Problem 6

Implement Circular Queue using an array.

````python
class CircularQueue:
    def __init__(self, capacity):
        self.capacity = capacity          # Maximum size of the queue
        self.queue = [None] * capacity    # Fixed-size array
        self.front = -1                   # Index of front element
        self.rear = -1                    # Index of rear element
        self.count = 0                    # Number of elements

    def enqueue(self, value):
        if self.is_full():
            raise Exception("Queue Full")

        # First element case
        if self.front == -1:
            self.front = 0

        # Move rear forward in circular manner
        self.rear = (self.rear + 1) % self.capacity
        self.queue[self.rear] = value
        self.count += 1

    def dequeue(self):
        if self.is_empty():
            raise Exception("Queue Empty")

        value = self.queue[self.front]
        self.queue[self.front] = None  # Optional: clear slot

        # If queue becomes empty after removal
        if self.front == self.rear:
            self.front = -1
            self.rear = -1
        else:
            # Move front forward in circular manner
            self.front = (self.front + 1) % self.capacity

        self.count -= 1
        return value

    def peek(self):
        if self.is_empty():
            raise Exception("Queue Empty")
        return self.queue[self.front]

    def is_empty(self):
        return self.count == 0

    def is_full(self):
        return self.count == self.capacity

    def size(self):
        return self.count

    def show(self):
        if self.is_empty():
            print("Queue is empty")
            return

        print("Queue contents:", end=" ")
        i = self.front
        for _ in range(self.count):
            print(self.queue[i], end=" ")
            i = (i + 1) % self.capacity
        print()
````

````python
# Create a circular queue of capacity 5
cq = CircularQueue(5)

# Enqueue elements
cq.enqueue(10)
cq.enqueue(20)
cq.enqueue(30)
cq.enqueue(40)
cq.show()   # Expected: 10 20 30 40

# Dequeue two elements
print("Dequeue:", cq.dequeue())  # Expected: 10
print("Dequeue:", cq.dequeue())  # Expected: 20
cq.show()   # Expected: 30 40

# Enqueue more elements (wrap around)
cq.enqueue(50)
cq.enqueue(60)
cq.show()   # Expected: 30 40 50 60

# Peek front
print("Peek front:", cq.peek())  # Expected: 30

# Enqueue last element to fill queue
cq.enqueue(70)
cq.show()   # Expected: 30 40 50 60 70

# Try enqueue when full
try:
    cq.enqueue(80)
except Exception as e:
    print("Error:", e)  # Expected: Queue Full

# Dequeue all elements
while not cq.is_empty():
    print("Dequeue:", cq.dequeue())

cq.show()   # Expected: Queue is empty
````

***Problem 7***

Why is a queue called a FIFO structure?

A queue is called a **FIFO (First In, First Out)** structure because the first element added to the queue is the first element removed. This is similar to people waiting in a line: the person who arrives first is served first. For example, if we enqueue `10`, then `20`, then 30, the order is `10 → 20 → 30`. When we perform `dequeue()`, `10` is removed first, followed by `20`, and then `30`. This behavior is what makes a queue different from a stack, which follows LIFO (Last In, First Out).

---

***Problem 8***

Why does a naive array queue suffer from memory bottlenecks?

A naive array queue can suffer from a memory bottleneck because when elements are removed from the front, the empty positions may not be reused efficiently. For example, if an array has a capacity of five and we add `10, 20, 30, 40, 50`, then remove `10, 20, 30`, the first three positions become empty, but the rear may already be at the end of the array. A simple implementation might consider the queue full even though there is unused space at the beginning. One solution is to shift all remaining elements toward the front, but this requires **$O(N)$** time. A circular queue solves this problem by allowing the rear to wrap around and reuse the empty positions.

---
***Problem 9***

Explain why a linked-list queue naturally supports O(1) enqueue and dequeue.

A linked-list queue naturally supports O(1) enqueue and dequeue when it maintains both a `front` pointer and a `rear` pointer. To enqueue an element, we create a new node and connect it directly after the current `rear`, then move the `rear` pointer to the new node, so no existing elements need to be moved. To dequeue an element, we simply move the `front` pointer to the next node and remove the old front node. Since both operations only modify a constant number of pointers regardless of how many elements are in the queue, both enqueue and dequeue take **$O(1)$** time.
